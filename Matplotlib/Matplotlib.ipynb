```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%importlib inline

plt.style.use('default')
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]

plt.plot(x,y)
plt.plot(y,x)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Simple Line Plot')
plt.show()
```

```python
#methord 2 object oriented approach 
fig, ax = plt.subplots() 
ax.plot(x,y)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Simple Line Plot(Object Oriented Approach)')
plt.show()
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)

plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.xlabel('X values')
plt.ylabel('Sin(x)')
plt.title('Sine Wave')
plt.grid(True)
plt.show()
```

```python
x=np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)
y3=np.sin(x)*np.cos(x)

plt.figure(figsize=(10,6))
plt.plot(x,y1,label='sin(x)')
plt.plot(x,y2,label='cos(x)')
plt.plot(x,y3,label='sin(x)*cos(x)')
plt.xlabel('X values')
plt.ylabel('Function values')
plt.title('Multiple Functions Plot')
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```

```python
np.random.seed(42)
x=np.random.randn(100)
y=2*x+np.random.randn(100)*0.5

plt.figure(figsize=(8,5))
plt.scatter(x,y,alpha=0.7) #alpha controls transparency
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter plot')
plt.grid(True,alpha=0.3)
plt.show()
```

```python
np.random.seed(42)
x=np.random.randn(100)
y=np.random.randn(100)
colors=np.random.randn(100)
size=1000*np.random.randn(100)

plt.figure(figsize=(10,6))
plt.scatter(x,y,c=colors,s=size,alpha=0.5,cmap='viridis')
plt.colorbar(label="color intensity")
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter Plot with Color and Size')
plt.grid(True,alpha=0.3)
plt.show()
```

```python
category=['A','B','C','D']
values=[2,7,8,9]

plt.figure(figsize=(10,6))
plt.bar(category,values,color='skyblue')
plt.xlabel('Categories')
plt.ylabel('values')
plt.title('Bar Chart')
plt.grid(True)
plt.show()
```
